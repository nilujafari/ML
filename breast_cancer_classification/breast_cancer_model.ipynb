```python
breast_cancer['diagnosis'] = breast_cancer['diagnosis'].map({'M': 1,'B': 0})
```

```python
data = breast_cancer.drop(['id', 'diagnosis'], axis= 1) 
target = breast_cancer['diagnosis']
```

# Test , Train

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(data, target, test_size= 0.2) #random_state= , stratify=
```

```python
print(f"X_train shape: {X_train.shape} - X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape} - y_test shape: {y_test.shape}")
```

# Histogram

```python
plt.figure(figsize=(4,4))
sns.countplot(x=y_train, hue=y_train, palette=["#4A6E55", "#804848"])
plt.title('Distribution of Target Variable')
plt.xlabel('Diagnosis (0=Benign, 1=Malignant)')
plt.ylabel('Count')
plt.show()
```

# Scaling (pre-processing(0-1))

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0,1))
# print(scaler.min_)
X_train = scaler.fit_transform(X_train)
# print(scaler.data_range_)
X_test  = scaler.transform(X_test)
```

# Binary Classification_report

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

def evaluation_metrics(y_train, y_pred_train, y_test, y_pred_test):
    accuracy_train = accuracy_score(y_train, y_pred_train)
    accuracy_test = accuracy_score(y_test, y_pred_test)
    precision = precision_score(y_test, y_pred_test)
    recall = recall_score(y_test, y_pred_test)
    f1 = f1_score(y_test, y_pred_test)
    print(f'Accuracy Train:{accuracy_train} - Accuracy Test:{accuracy_test} - Precision:{precision} - Recall:{recall} - F1_Score:{f1}')

    print('\nConfusion_Matrix:')
    print(confusion_matrix(y_test, y_pred_test))
    
    print('\nClassification_Report:')
    print(classification_report(y_test, y_pred_test))
    
    return accuracy_train, accuracy_test, precision, recall, f1
```

# Model training

## 1. KNN (K Nearest Neighbors)

```python
from sklearn.neighbors import KNeighborsClassifier

scores = []
for k in range(13,17):
    knn = KNeighborsClassifier(n_neighbors= k)
    knn.fit(X_train, y_train)
    score = knn.score(X_test, y_test)
    scores.append(score)
    print(f'K = {k}, Accuracy = {score}')
```

```python
knn = KNeighborsClassifier(n_neighbors= 13, weights='uniform', metric='euclidean')
knn.fit(X_train, y_train)
```

```python
y_pred_train = knn.predict(X_train)
y_pred_test = knn.predict(X_test)

accuracy_train_knn, accuracy_test_knn, precision_knn, recall_knn, f1_knn = evaluation_metrics(y_train, y_pred_train, y_test, y_pred_test)
```

```python
import seaborn as sns
confusion_m = confusion_matrix(y_test, y_pred_test)
plt.figure(figsize=(3,2))
sns.heatmap(confusion_m, annot=True, fmt ='d', cmap='hot')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix KNN')
plt.show()
```

## 2. Decition Tree

```python
from sklearn.tree import DecisionTreeClassifier

best_score = 0
best_parameters = []
for depth in [2, 3, 4, 5, 6]:
    for leaf in [4, 5, 6, 7, 8]:
        for split in [2, 5, 8, 10, 12]:
            dt = DecisionTreeClassifier(max_depth=depth, min_samples_leaf=leaf, min_samples_split=split)
            dt.fit(X_train, y_train)
            y_pred_test = dt.predict(X_test)
            score = accuracy_score(y_test, y_pred_test)
            # print(f'max_depth={depth}, min_samples_leaf={leaf}, min_samples_split={split}, accuracy={score}')

            if score > best_score:
                best_score = score
                best_parameters = {"max_depth":depth, "min_samples_leaf":leaf, "min_samples_split":split}
print('\nBest Accuracy:', best_score)
print('Best Parameters:', best_parameters)
```

```python
dt = DecisionTreeClassifier(max_depth= 5, min_samples_leaf=4, min_samples_split=10)
dt.fit(X_train, y_train)
```

```python
y_pred_train = dt.predict(X_train)
y_pred_test = dt.predict(X_test)

accuracy_train_dt, accuracy_test_dt, precision_dt, recall_dt, f1_dt = evaluation_metrics(y_train, y_pred_train, y_test, y_pred_test)
```

```python
import seaborn as sns
confusion_m = confusion_matrix(y_test, y_pred_test)
plt.figure(figsize=(3,2))
sns.heatmap(confusion_m, annot=True, fmt ='d', cmap='hot')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix Decision Tree')
plt.show()
```

## 3. Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

best_score = 0
best_parameters = []
for n in [100, 150, 200]:
    for depth in [5, 10, 15]:
        rf = RandomForestClassifier(n_estimators=n, max_depth=depth)
        rf.fit(X_train, y_train)
        y_pred_test = rf.predict(X_test)
        score = accuracy_score(y_test, y_pred_test)
        print(f'n_estimators={n}, max_depth={depth}, accuracy={score}')

        if score > best_score:
            best_score = score
            best_parameters = {"n_estimators":n, "max_depth":depth}
print('\nBest Accuracy:', best_score)
print('Best Parameters:', best_parameters)
```

```python
rf = RandomForestClassifier(n_estimators=100, max_depth=5, max_features='sqrt')
rf.fit(X_train, y_train)
```

```python
y_pred_train = rf.predict(X_train)
y_pred_test = rf.predict(X_test)

accuracy_train_rf, accuracy_test_rf, precision_rf, recall_rf, f1_rf = evaluation_metrics(y_train, y_pred_train, y_test, y_pred_test)
```

```python
import seaborn as sns
confusion_m = confusion_matrix(y_test, y_pred_test)
plt.figure(figsize=(3,2))
sns.heatmap(confusion_m, annot=True, fmt ='d', cmap='hot')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix Random Forest')
plt.show()
```

```python
y_pred_train = svm.predict(X_train)
y_pred_test = svm.predict(X_test)

accuracy_train_svm, accuracy_test_svm, precision_svm, recall_svm, f1_svm = evaluation_metrics(y_train, y_pred_train, y_test, y_pred_test)
```

```python
import seaborn as sns
confusion_m = confusion_matrix(y_test, y_pred_test)
plt.figure(figsize=(3,2))
sns.heatmap(confusion_m, annot=True, fmt ='d', cmap='hot')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix SVM')
plt.show()
```

## 6. ANN (Artificial Neural Network)

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(hidden_layer_sizes=800, max_iter=600, activation='relu', solver='adam')
mlp.fit(X_train, y_train)
```

```python
y_prob = mlp.predict_proba(X_test)[:,1]
for threshold in [0.5, 0.4, 0.3, 0.2]:
    y_pred_test = (y_prob > threshold).astype(int)
    print("Threshold:", threshold, "Recal:", recall_score(y_test, y_pred_test))
```

```python
threshold = 0.3
y_pred_test = (y_prob > threshold).astype(int)

accuracy_train_mlp, accuracy_test_mlp, precision_mlp, recall_mlp, f1_mlp = evaluation_metrics(y_train, y_pred_train, y_test, y_pred_test)
```

```python
import seaborn as sns
confusion_m = confusion_matrix(y_test, y_pred_test)
plt.figure(figsize=(3,2))
sns.heatmap(confusion_m, annot=True, fmt ='d', cmap='hot')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix ANN')
plt.show()
```
