# Typing

```python
from typing import List, Dict, Tuple, Any
```

```python
def func1(name:str, age: int, gpa: float, alive: bool):
    print(f"Customer name is: {name}, age: {age}, gpa:{gpa}, alive:{alive}")
```

```python
func1(name="Ali", age=15, gpa=18.43, alive=True)
```

```python
def funct2(data: Dict[str, Any]):
    print(f"Customer name is: {data['name']}, age: {data['age']}, gpa:{data['gpa']}, alive:{data['alive']}")
```

```python
funct2({
    "name":"Ali",
    "age": 15,
    "gpa": 13.43,
    "alive": True
})
```
