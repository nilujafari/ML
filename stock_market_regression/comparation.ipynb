```python
import json
import matplotlib.pyplot as plt

files = {
    "Main_amz": "Main_amz_plot.json",
    "Not Shuffle": "not_shuffle_plot.json",
    "14 Days": "pred_fourteenday_plot.json",
    "7 Days": "pred_sevenday_plot.json"}
all_data = {}
for name, file in files.items():
    with open(file, "r") as f:
        all_data[name] = json.load(f)
```

```python
fig, ax = plt.subplots(4, 3, figsize=(15, 16))

for row, (name, data) in enumerate(all_data.items()):

    models = data["models"]

    ax[row, 0].bar(models, data["MAE"])
    ax[row, 0].set_title(f"{name} - MAE")
    ax[row, 0].tick_params(axis="x", rotation=45)
    ax[row, 0].grid(axis="y")

    ax[row, 1].bar(models, data["MSE"])
    ax[row, 1].set_title(f"{name} - MSE")
    ax[row, 1].tick_params(axis="x", rotation=45)
    ax[row, 1].grid(axis="y")

    ax[row, 2].bar(models, data["RMSE"])
    ax[row, 2].set_title(f"{name} - RMSE")
    ax[row, 2].tick_params(axis="x", rotation=45)
    ax[row, 2].grid(axis="y")

plt.tight_layout()
plt.show()
```
